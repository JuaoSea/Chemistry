```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

arquivo = r"iv2_no.txt"

dados = np.loadtxt(arquivo, comments='#')

x = dados[:, 0]   # número de onda (cm^-1)
y = dados[:, 1]   # transmitância (%)

# Detectar mínimos como picos
picos, propriedades = find_peaks(-y, prominence=6)

x_picos = x[picos]
y_picos = y[picos]


plt.figure(figsize=(10, 6))
plt.plot(x, y, color='black', linewidth=1.2)
plt.scatter(x_picos, y_picos, color='red', s=5)

for xp, yp in zip(x_picos, y_picos):
    plt.text(
        xp,
        yp - 1,
        f'{xp:.0f}',
        fontsize=8,
        rotation=90,
        ha='center',
        va='top'
    )

plt.xlabel(r'Número de onda (cm$^{-1}$)', fontsize=12)
plt.ylabel('Transmitância (%)', fontsize=12)

plt.title(r'Espectro IV - $[Co(NH$_3$)_5(NO_2)]Cl_2$', fontsize=13)

plt.xlim(4000, 400)
plt.ylim(40, 105)

plt.grid(False)

plt.tight_layout()
#plt.savefig('iv_spectrum_no.png', dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

arquivo1 = r"iv2_no.txt"
arquivo2 = r"iv2_ono.txt"

dados1 = np.loadtxt(arquivo1, comments='#')
dados2 = np.loadtxt(arquivo2, comments='#')

x1 = dados1[:, 0]
y1 = dados1[:, 1]

x2 = dados2[:, 0]
y2 = dados2[:, 1]

# Espectro 1
picos1, _ = find_peaks(-y1, prominence=11)

x_picos1 = x1[picos1]
y_picos1 = y1[picos1]

# Espectro 2
picos2, _ = find_peaks(-y2, prominence=6)

x_picos2 = x2[picos2]
y_picos2 = y2[picos2]

plt.figure(figsize=(11, 6))

# Espectro 1
plt.plot(
    x1,
    y1,
    color='royalblue',
    linewidth=1.5,
    label=r'$[Co(NH_3)_5(NO_2)]Cl_2$'
)

# Espectro 2
plt.plot(
    x2,
    y2,
    color='darkorange',
    linewidth=1.5,
    label=r'$[Co(NH_3)_5(ONO)]Cl_2$'
)

plt.scatter(
    x_picos1,
    y_picos1,
    color='blue',
    s=10
)

for xp, yp in zip(x_picos1, y_picos1):
    plt.text(
        xp,
        yp - 1,
        f'{xp:.0f}',
        fontsize=7,
        rotation=90,
        ha='center',
        va='top',
        color='blue'
    )

plt.scatter(
    x_picos2,
    y_picos2,
    color='darkorange',
    s=10
)

for xp, yp in zip(x_picos2, y_picos2):
    plt.text(
        xp,
        yp - 4,
        f'{xp:.0f}',
        fontsize=7,
        rotation=90,
        ha='center',
        va='bottom',
        color='darkorange'
    )

plt.xlabel(r'Número de onda (cm$^{-1}$)', fontsize=12)
plt.ylabel('Transmitância (%)', fontsize=12)

plt.title(
    r'Espectros IV - Isômeros Nitro e Nitrito',
    fontsize=14
)

plt.xlim(4000, 400)
plt.ylim(30, 105)
plt.legend()
plt.grid(False)
plt.tight_layout()
#plt.savefig('iv_duplo.png', dpi=300)
plt.show()
```
