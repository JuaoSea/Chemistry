# Lei de PLanck

Vamos apresentar brevemente os principais aspectos da teoria combinando alguns artigos da Wikipédia. Links abaixo.

A lei de Planck descreve a densidade espectral da radiação eletromagnética emitida por um corpo negro em equilíbrio térmico a uma dada temperatura, quando não há fluxo líquido de matéria ou energia entre o corpo e seu ambiente.

A lei de Planck pode ser escrita em termos da densidade espectral de energia. Essas distribuições têm unidades de energia por volume por unidade espectral

$\rho (\lambda, T) = \frac{8 \pi h c}{\lambda^5 \left(e^{\frac{hc}{\lambda k_B T}} - 1\right)}$

Onde $k_B$ é a constante de Boltzmann, $h$ é a constante de Planck, e $c$ é a velocidade da luz no meio, seja material ou vácuo.

A radiação de corpo negro é a radiação eletromagnética térmica dentro ou ao redor de um corpo em equilíbrio termodinâmico com seu ambiente, emitida por um corpo negro (um corpo opaco e não reflexivo idealizado). Ele tem um espectro específico de comprimentos de onda, inversamente relacionado à intensidade que depende apenas da temperatura do corpo, que é assumido para fins de cálculos e teoria como uniforme e constante.

Referências: [Lei de Planck](https://en.wikipedia.org/wiki/Planck%27s_law), [Radiação do corpo negro](https://en.wikipedia.org/wiki/Black-body_radiation), [Densidade de energia](https://en.wikipedia.org/wiki/Energy_density)


```python
import numpy as np

from ipywidgets import interactive, fixed

from planck import Planck
```

```python
lambda_array = np.linspace(1.0e-9, 2.0e-6, 1000)
temperature_array = np.arange(1000, 7001, 500)
example = Planck(lambda_array, temperature_array)
example.plot()
```

```python
import matplotlib.pyplot as plt

example.plot(plt.cm.gist_rainbow)
```

```python
example.plot(plt.cm.gist_rainbow, lines=200, transparency=0.1)
```

```python
example_one_temperature = Planck(lambda_array, (3000,))
example_one_temperature.plot()
```

```python
example.plot_visible()
```

```python
example.plot_visible(lines=5000)
```

```python
graph = interactive(Planck.plot_interactive,
                    wavelengths=fixed(lambda_array),
                    transparency=fixed(0.1),
                    ax=fixed(None),
                    temperature=(3000,10000,100))
display(graph)
```
