# ⚛️ Chemistry Visualizer

Este notebook permite visualizar e analisar moléculas a partir de SMILES, utilizando RDKit, pandas e py3Dmol para visualização 2D e 3D.

---

## 🔍 Funcionalidades

- **Transformação de SMILES em moléculas RDKit**
- **Visualização 2D** com fórmulas moleculares c
- **Visualização 3D** interativa de cada molécula
- **Fácil customização** da lista de moléculas

---

## 🛠️ Como usar

1. **Edite a lista de SMILES** na variável `smiles_exemplo`.
2. **Execute as células** para gerar visualizações 2D e 3D das moléculas.
3. **Explore as estruturas** diretamente no notebook.

---

## 📚 Requisitos

- [RDKit](https://www.rdkit.org/)
- [pandas](https://pandas.pydata.org/)
- [py3Dmol](https://3dmol.csb.pitt.edu/)

> ⚠️ Certifique-se de que todas as bibliotecas estejam instaladas no seu ambiente Python.

---

## 💡 Exemplos de uso

- Visualização rápida de moléculas a partir de SMILES.
- Geração automática de fórmulas moleculares em formato químico.
- Inspeção interativa de estruturas em 3D.

---

```python
from rdkit import Chem
from rdkit.Chem import Draw, Descriptors, Crippen, AllChem, rdMolDescriptors
from IPython.display import display, HTML
import py3Dmol
import pandas as pd
import re
```

```python
smiles_exemplo = ["CN1C=NC2=C1C(=O)N(C(=O)N2C)C", "O=C(Oc1ccccc1C(=O)O)C", "C(C(C1C(=C(C(=O)O1)O)O)O)O", "NCCc1c[nH]c2ccc(O)cc12"]
```

```python
def formatar_formula(formula):
    # Substitui números por subscritos HTML
    return re.sub(r'(\d+)', r'<sub>\1</sub>', formula)
 
mols = [Chem.MolFromSmiles(smiles) for smiles in smiles_exemplo]
 
legendas = [f"{formatar_formula(rdMolDescriptors.CalcMolFormula(mol))}" for i, mol in enumerate(mols)]
 
# Desenhar múltiplas moléculas com fundo branco forçado no SVG
img = Draw.MolsToGridImage(mols, molsPerRow=len(smiles_exemplo), subImgSize=(300, 300), legends=legendas, useSVG=True)
svg = img.data if hasattr(img, 'data') else img._repr_svg_()
svg = svg.replace('<svg', '<svg style="background:white"', 1)
display(HTML(svg))
```

```python
# Visualizar todas as moléculas de smiles_exemplo em 3D com py3Dmol
for i, smiles in enumerate(smiles_exemplo):
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol)
    AllChem.EmbedMolecule(mol)
    AllChem.MMFFOptimizeMolecule(mol)
    view = py3Dmol.view(width=350, height=350)
    view.addModel(Chem.MolToMolBlock(mol), 'mol')
    view.setStyle({'stick': {}})
    view.setBackgroundColor('0xeeeeee')
    view.zoomTo()
    print(f"Molécula {i+1} - SMILES: {smiles}")
    view.show()
```
